```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data=pd.read_csv('housing.csv')
```

```python
data.info()
```

```python
data = data.dropna()
```

```python
data
```

```python
data = pd.get_dummies(data)
data
```

```python
data['bedroom_ratio'] = data['total_bedrooms']/data['total_rooms']
data
```

```python
data['household_rooms']=data['total_rooms']/data['households']
data
```

```python
Datas = data
Datas['population']=np.log(Datas['population']+1)
Datas['households']=np.log(Datas['households']+1)
Datas['total_bedrooms']=np.log(Datas['total_bedrooms']+1)
Datas['total_rooms']=np.log(Datas['total_rooms']+1)
Datas
```

```python
Datas['household_rooms']=np.log(Datas['household_rooms']+1)
Datas
```

```python
#from sklearn.model_selection import train_test_split

x = Datas.drop(['median_house_value'],axis=1)
y = Datas['median_house_value']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression


reg=LinearRegression()

reg.fit(x_train, y_train)
```

```python
reg.score(x_test,y_test)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scaler=scaler.fit_transform(x_train)
x_test_scaler=scaler.fit_transform(x_test)
```

```python
reg.fit(x_train_scaler,y_train)
```

```python
reg.score(x_test_scaler,y_test)
```

```python
from sklearn.ensemble import RandomForestRegressor

forest = RandomForestRegressor()
```

```python
forest.fit(x_train,y_train)
```

```python
forest.score(x_test,y_test)
```

```python
importances = forest.feature_importances_

# 4. Organisation dans un DataFrame Pandas pour trier facilement
df_importance = pd.DataFrame({
    'Feature': x.columns,
    'Importance': importances
}).sort_values(by='Importance', ascending=False) # Tri du plus important au moins important

# Affichage du tableau textuel
print(df_importance)

# 5. BONUS : Visualisation graphique avec Matplotlib
plt.figure(figsize=(10, 6))
plt.barh(df_importance['Feature'], df_importance['Importance'], color='skyblue')
plt.gca().invert_yaxis() # Pour avoir la feature la plus importante en haut
plt.xlabel('Score d\'Importance')
plt.title('Importance des Features - Random Forest')
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV

forest= RandomForestRegressor()
param_grid={
    'n_estimators':[250,420,560],
    'max_features':[5,7,10]
}

grid_search = GridSearchCV(forest, param_grid=param_grid, cv=5,n_jobs=4, scoring='neg_mean_squared_error')
grid_search.fit(x_train,y_train)
```

```python
grid_search.best_estimator_
```

```python
best_forest_model = grid_search.best_estimator_
best_forest_model.score(x_test,y_test)
```

```python
### Essayer de predire le prix d'une maison :
### Firest Essay
### au debut, on creer une liste de categories, afin de pouvoir les convertir en False, True, de facon que sa 
## soit compatible avec le modele de prediction : Random Forest ( best_forest_model)


import numpy as np
import pandas as pd

# 1. Définition des catégories du dataset original (d'après votre image)
categories_ocean = [
    '<1H OCEAN',
    'INLAND',
    'ISLAND',
    'NEAR BAY',
    'NEAR OCEAN',
]  #[cite: 1]

# 2. Votre DataFrame initial
Nouvelle_colonne = pd.DataFrame([
    {
        'longitude': -122,
        'latitude': 37.24,
        'housing_median_age': 33,
        'total_rooms': 2840,
        'total_bedrooms': 560,
        'population': 1890,
        'households': 489,
        'median_income': 7.14,
        'ocean_proximity': 'INLAND',
    }
])

# Transformations log et features
Nouvelle_colonne['bedroom_ratio'] = (
    Nouvelle_colonne['total_bedrooms'] / Nouvelle_colonne['total_rooms']
)
Nouvelle_colonne['household_rooms'] = (
    Nouvelle_colonne['total_rooms'] / Nouvelle_colonne['households']
)
Nouvelle_colonne['total_rooms'] = np.log(Nouvelle_colonne['total_rooms'] + 1)
Nouvelle_colonne['total_bedrooms'] = np.log(
    Nouvelle_colonne['total_bedrooms'] + 1
)
Nouvelle_colonne['households'] = np.log(Nouvelle_colonne['households'] + 1)
Nouvelle_colonne['population'] = np.log(Nouvelle_colonne['population'] + 1)

# 3. Forcer les catégories possibles avant le get_dummies
Nouvelle_colonne['ocean_proximity'] = pd.Categorical(
    Nouvelle_colonne['ocean_proximity'], categories=categories_ocean
)

# 4. Encodage One-Hot (génèrera automatiquement les 5 colonnes)
Nouvelle_colonne = pd.get_dummies(
    Nouvelle_colonne, columns=['ocean_proximity']
)

# 5. S'assurer que l'ordre des colonnes correspond exactement au modèle (Optionnel mais recommandé)
# Nouvelle_colonne = Nouvelle_colonne[model.feature_names_in_
```

```python
### creons une fonction pour predire directement depuis l'input de l'User..

a = float(input('Donner la Longitude du maison :'))
b = float(input('Donner la Latitude du maison :'))
c = int(input('Donner la moyenne a peu pres, de l\'age des gens occupant cette quartier :'))
d = int(input('Donner le nombre total des salles existant dans ce quartier :'))
e = int(input('Donner le nombre total des chambres existant dans ce quartier :'))
f = int(input('Donner la population de cette quartier :'))
g = int(input('Donner le nombre a peu pres, des personnes occupant un appartement dans la zone du maison a predire\n:'))
h = float(input('Donner le revenu moyenne des residents par an (Dollars) :'))
print('Pour la question final, je vous demande de me repondre par un chiffre 1 - 5 \n:')
i=int(input('La zone geographique du maison par rapport a la mer :1-La zone est située à moins de 60 minutes de route de la côte océanique.\n 2-La zone se trouve à l\'intérieur des terres, éloignée de la côte et de l\'océan.\n 3-La zone se trouve très près de la côte océanique (souvent dans la bande côtière immédiate).\n 4-La zone est située à proximité d\'une baie (principalement la baie de San Francisco).\n 5-Le logement est situé sur une île (par exemple, l\'île de Santa Catalina au large de la Californie).\n'))
match i:
    case 1:
        i='<1H OCEAN'
    case 2:
        i='INLAND'
    case 3:
        i='NEAR OCEAN'
    case 4:
        i='NEAR BAY'
    case 5:
        i='ISLAND'
    case _:
        raise ValueError("Votre reponse est indefinie....")
def predict_house_price(a,b,c,d,e,f,g,h,i):
    global best_forest_model
    
    col = pd.DataFrame([
        {
            'longitude': a,
            'latitude': b,
            'housing_median_age': c,
            'total_rooms': d,
            'total_bedrooms': e,
            'population': f,
            'households': g,
            'median_income': h,
            'ocean_proximity': i,
        }
    ])
    
    # Transformations log et features
    col['bedroom_ratio'] = (
        col['total_bedrooms'] / col['total_rooms']
    )
    col['household_rooms'] = (
        col['total_rooms'] / col['households']
    )
    col['total_rooms'] = np.log(col['total_rooms'] + 1)
    col['total_bedrooms'] = np.log(
        col['total_bedrooms'] + 1
    )
    col['households'] = np.log(col['households'] + 1)
    col['population'] = np.log(col['population'] + 1)
    
    # 3. Forcer les catégories possibles avant le get_dummies
    col['ocean_proximity'] = pd.Categorical(
        col['ocean_proximity'], categories=categories_ocean
    )
    
    # 4. Encodage One-Hot (génèrera automatiquement les 5 colonnes)
    col = pd.get_dummies(
        col, columns=['ocean_proximity']
    )
    col = col[best_forest_model.feature_names_in_]
    valeure_predite = best_forest_model.predict(col)[0]
    return round(valeure_predite,2)

valeure_predite=predict_house_price(a,b,c,d,e,f,g,h,i)
print("Le prix predite du votre maison ( L'IA est un modele, il peut se tromper, soyez sure que c'est a 84 % correct): ",valeure_predite)
```
